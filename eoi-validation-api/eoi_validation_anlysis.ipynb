### SBM Database Fild For Underwritten Life & Disability

### Where :

- ULD -> Underwriter Life Disability

- LIFE_DI -> Life Disability Insurance

- ULD_SUBGROUPS → subgroup information

- ULD_BILLING_ENROLL → billing/enrollment data

- ULD_CONTACTS → contact details

- ULD_SOLD_RATES → sold/premium rate information


## Table Codes : Suffix Description
----------------------------------------------------
|  Suffix  | Description                           |
|----------|---------------------------------------|
|BL        | Basic Life                            |
|SL        | Supplemental Life                     |
|DL        | Dependent Life                        |

|SDLS      | Supplemental Dependent Life Spouse    |

|BADD      | Basic AD&D                            |
|SADD      | Supplemental AD&D                     |
|DADD      | Dependent AD&D                        |
|SDADD     | Supplemental Dependent AD&D           |
|STD       | Short Term Disability                 |
|STDBU     | STD Buy-Up                            |
|LTD       | Long Term Disability                  |
|LTDBU     | LTD Buy-Up                            |
|GI        | General Information / Guaranteed Issue|

## Mapping Fild of the TBM Table ULD_LIFE_DI

```python
sql_query= """SELECT TS_ID,TS_GROUP_POLICY_NUMBER AS GROUP_NUMBER, TS_GROUP_NAME AS EMPLOYER_NAME,
TS_SITUS_STATE AS SITUS_STATE,
TS_PLAN_ANNIV_DATE AS PLAN_ANNIVERSARY_DATE,
TS_GUAR_ISSUE_AMT_BL AS BASIC_LIFE_GI_AMOUNT,
TS_GUAR_ISSUE_AMT_SL AS SUPP_LIFE_GI_AMOUNT,
TS_BEN_PERCENT_LTD AS LTD_BENEFIT_PERCENTAGE,
TS_BENEFIT_PERCENTAGE_STD AS STD_BENEFIT_PERCENTAGE,
TS_MAX_MONTH_BEN_LTD AS MAXIMUM_MONTHLY_BENEFIT,
TS_MAX_WEEK_BENE_STD AS MAXIMUM_WEEKLY_BENEFIT,
TS_OTHER_SP_SL AS SPOUSE_ELIGIBILITY_COMMENT,
TS_EXP_LIMIT_SDL_SPOUSE AS SPOUSE_COVERAGE_LIMIT_COMMENT,
TS_COMMENTS_BL AS BASIC_LIFE_COMMENTS,
TS_COMMENTS_SL AS SUPP_LIFE_COMMENTS,
TS_COMMENTS_LTD AS LTD_COMMENTS,
TS_COMMENTS_STD AS STD_COMMENTS
FROM ULD_LIFE_DI"""

map_df = pd.read_sql(sql_query,engine)
```
